**Object oriented Programming: Class and Objects**

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed


    def sit(self, place):
        return f"{self.name} is sitting {place}."
    
dog1 = Dog("Tommy", "German Sephard")
dog1.sit("on the sofa")
print(dog1.sit("on the sofa"))
```

**File I/O**

```python
with open("example.txt", "w") as f:
    f.write("Title: Example text file\n")
    text = input("Enter the text you want to write or \"no\" to stop writing:")
    if text == "no":
        exit
    else:
        f.write(text)
```

```python
with open("example.txt", "r") as f:
    content = f.readlines()
    print(content)
```

```python
with open("example.txt", "a") as f:
    flag = True
    while flag == True:
        text = input("Enter the text you want to write or \"no\" to stop writing:")
        if text == "no":
            flag = False
        else:
            f.write("\n" + text)
```

**Collections, Iterators, Generators and Decorators**

```python
# Collections module has data structures like deque, counter, defaultdict, namedtuple etc

from collections import Counter

aList = [1, 1, 2, 3, 1, 4, 4]
Counter(aList)

from collections import defaultdict

e = {'a': 1, 'b': 2}
d = defaultdict(int)
d['a'], d['b'] = 1, 2
print(d)
print(e)
print(d['c'])
# print(e['c']) # With normal dict it gives key error when we access dictionary with wrong key

from collections import deque

double_ended_queue = deque([1, 2, 3])
print(double_ended_queue)
double_ended_queue.appendleft(0)
double_ended_queue.append(4)
print(double_ended_queue)

from collections import namedtuple

named_tuple = namedtuple("Object_lite", "x y")
n1 = named_tuple(2, 3)
print(n1)
print(f"X co-ordinate is {n1.x}")
```

**Iterators:
    Loops through item one by one**

```python
aList = [1, 1, 2, 3, 1, 4, 4]
it = iter(aList)

# print(next(it)) shows next element, when index is exceeded StopIteration error is displayed

# Iterators are what work under the hood of for loops
while True:
    try:
        print(next(it), end=" ")
    except StopIteration:
        break

print("\nNow with for loop:")

for i in aList:
    print(i, end=" ")
```

**Generators: Produce value only when needed, iterator with yield**

```python
def cubes(a):
    for i in range(a):
        yield i*i*i

for x in cubes(5):
    print(x)

def square(a):
    yield a*a

print(square(25))
```

**Decorators: Add more layers to a func()**

```python
def decorator(func):
    def wrapper(*args, **kwargs):   # accept any arguments
        print("Before the function.")
        func(*args, **kwargs)       # pass them along
        print("After the function.")
    return wrapper

@decorator
def cubes(a):
    print(f"{a*a*a} is the cube of {a}")

cubes(5)
```

**Functions and Lambda Expressions**

```python
# Simple one line functions for arithmetics

divide = lambda a, b: a / b
print(divide(10, 5))
```

```python
# For sorting 
aList = [("Ram", 45), ("Hari", 34), ("Sita", 23), ("Manav", 12)]
sortedList = sorted(aList, key= lambda x: x[1])
print(sortedList)
```

```python
# For filtering
aList = [("Ram", 45), ("Hari", 34), ("Sita", 23), ("Manav", 12)]
above30 = list(filter(lambda x: x[1] > 30, aList))
print(above30)
```

```python
# For mapping
names = ["Ram", "Sita", "Hari", "Om shanti om", "Bahadur", "Okil", "Otak", "Sparrowhawk", "Tenar"]
lastDigit = list(map(lambda x: x[-1], names))
lastDigit = sorted(lastDigit, key= lambda x: x[0])
print(lastDigit)
```
